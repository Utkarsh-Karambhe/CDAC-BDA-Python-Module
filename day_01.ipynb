### Exercise 1: Leap Year Checker

```python
year = int(input("Enter the year: "))

if((year%400==0) or (year%4==0 and year%100!=0)):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")

```

### Exercise 2: Fibonacci Sequence Generator

```python
n = int(input("Enter the range: "))

a, b = 0,1

for i in range(n):
    print(a,end =" ")
    c = a+b
    a=b
    b=c
```

### Exercise 3: Prime Number Checker

```python
n = int(input("Enter the number: "))

limit = n//2

if(n>1):
    for i in range(2,limit+1):
        if(n%i==0):
            print("Not Prime")
            break
    else:
        print("Prime")
else:
    print("Not Prime")

```

### Exercise 4: Odd or Even Checker

```python
num = int(input("Enter the number"))

if(num%2==0):
    print(f"{num} is even.")
else:
    print(f"{num} is odd.")
```

### Exercise 5: Basic Operator Calculator

```python
n1 = int(input("Enter the first num: "))
n2 = int(input("Enter the second num"))

operator = input("Enter the operator: ")

if(operator == "+"):
    print(n1+n2)
elif(operator == "-"):
    print(n1-n2)
elif(operator == "*"):
    print(n1*n2)
elif(operator == "/"):
    print(round(n1/n2,2))
else:
    print("Please enter the valid operator")
```

### Exercise 6: Sum of N Natural Numbers

```python
n = int(input("Enter the number of numbers you wish to add."))

sum = n*(n+1)/2

print(sum)
```

### Exercise 7: Multiplication Table Generator

```python
num = int(input("Enter the number who's table u wish to generate: "))

for i in range(1,11):
    print(f"{num} * {i} = {num*i}")
```

### Exercise 8: Score to Grade Converter

```python
score = int(input("Enter your score: "))

if(score<=100 and score>=90):
    print("Grade: A")
elif(score<=89 and score>=80):
    print("Grade: B")
elif(score<=79 and score>=70):
    print("Grade: C")
elif(score<=69 and score>=60):
    print("Grade: D")
else:
    print("Grade: F")
```
