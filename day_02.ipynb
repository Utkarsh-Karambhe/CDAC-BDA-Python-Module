### Exercise 4: Vowel & Consonant Frequency

```python
user_input = input("Enter the string: ")
a, e, i, o, u = 0,0,0,0,0
consonent = 0
for ch in user_input:

    if(ch==" "):
        continue
    else:
        if(ch=="a"):
            a+=1
        elif(ch=="e"):
            e+=1
        elif(ch=="i"):
            i+=1
        elif(ch=="o"):
            o+=1
        elif(ch=="u"):
            u+=1
        else:
            consonent+=1
    

print("Vowel Frequencies: \n")
print(f"a:{a}")
print(f"e:{e}")
print(f"i:{i}")
print(f"o:{o}")
print(f"u:{u}")
print(f"Total consonents: {consonent}")
```

### Exercise 6: Shift Cipher Encrypter

```python
txt = input("Enter the string: ")
result = ""

for ch in txt:
    converted = ord(ch)
    converted+=3
    convert_back = chr(converted)
    result = result+convert_back

print(result)
```

### Exercise 9: Longest Palindromic Substring

```python
txt = input("Enter the string: ")

longest = ""

for i in range(len(txt)):
    for j in range(i+1, len(txt)):
        sub = txt[i:j]

        if(sub==sub[::-1]):
            if(len(sub)>len(longest)):
                longest = sub

print(longest)
```

### Exercise 10: Run-Length String Compression

```python
txt = input("Enter the string: ")
result = ""
count = 1

for i in range(len(txt)-1):
    if(txt[i]==txt[i+1]):
        count+=1
    else:
        result += txt[i]+str(count)
        count = 1

result += txt[i]+str(count)

if(len(result)<len(txt)):
    print(result)
else:
    print(txt)
```

### Exercise 11: Group Anagrams

```python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]

groups = []

for word in words:
    found_group = False
    
    for group in groups:
        if sorted(word) == sorted(group[0]):
            group.append(word)
            found_group = True
            break  
    
    if not found_group:
        groups.append([word])

print(groups)
```

### Exercise 12: Date Validator & Pretty Formatter

```python

date_str = input("Enter date (DD/MM/YYYY): ")

parts = date_str.split("/")
day = int(parts[0])
month = int(parts[1])
year = int(parts[2])

if month < 1 or month > 12:
    print("Invalid Date")
else:
    if month == 2: 
 
        if year % 400 == 0:
            days_in_month = 29
        elif year % 100 == 0:
            days_in_month = 28
        elif year % 4 == 0:
            days_in_month = 29
        else:
            days_in_month = 28
    elif month == 4 or month == 6 or month == 9 or month == 11:
      
        days_in_month = 30
    else:
     
        days_in_month = 31
  
    if day < 1 or day > days_in_month:
        print("Invalid Date")
    else:

        month_names = ("January", "February", "March", "April", "May", "June",
                      "July", "August", "September", "October", "November", "December")
    
        month_name = month_names[month - 1]
        
        print(f"{month_name} {day}, {year}")
```
