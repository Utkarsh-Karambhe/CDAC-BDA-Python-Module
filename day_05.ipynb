### Assignment 1: CDAC Cafeteria Discount Calculator

```python
def calculate_cafeteria_bill(base_price, *items, tax_rate=0.05, discount=0.0, delivery_fee=0.0):

    raw_subtotal = base_price

    for item in items:
        raw_subtotal = raw_subtotal + item
   
    discount_decimal = discount / 100.0
    
 
    discounted_subtotal = raw_subtotal * (1 - discount_decimal)
    
    tax_amount = discounted_subtotal * tax_rate
    
    final_bill = discounted_subtotal + tax_amount + delivery_fee
    
    final_bill = round(final_bill, 2)

    return final_bill


total1 = calculate_cafeteria_bill(100.0)
print(f"Test 1: ${total1}")  

total2 = calculate_cafeteria_bill(100.0, 20.0, 30.0, tax_rate=0.08, discount=10.0, delivery_fee=15.0)
print(f"Test 2: ${total2}")  

total3 = calculate_cafeteria_bill(50.0, tax_rate=0.10, discount=5.0, delivery_fee=10.0)
print(f"Test 3: ${total3}")

total4 = calculate_cafeteria_bill(100.0, 10.0, 15.0, 20.0, 25.0)
print(f"Test 4: ${total4}")
```

### Assignment 3: Corporate Directory Search & Scraper

```python
import re

def scrape_directory_phones(directory_text):    
    pattern = r'(\d{3})[-)]?(\d{3})[- ]?(\d{4})'
    
    matches = re.findall(pattern, directory_text)
    
    results = []
    
    for match in matches:
  
        area_code = match[0]
        prefix = match[1]
        line_number = match[2]
     
        phone_info = {
            "area_code": area_code,
            "prefix": prefix,
            "line_number": line_number,
            "formatted": "(" + area_code + ") " + prefix + "-" + line_number
        }
      
        results.append(phone_info)
    
    return results

directory = "Contact HR at 123-456-7890 or the helpdesk at (987) 654-3210. Direct line is 5558881234."
result = scrape_directory_phones(directory)

print("Found phone numbers:")
for phone in result:
    print(f"  Area Code: {phone['area_code']}")
    print(f"  Prefix: {phone['prefix']}")
    print(f"  Line Number: {phone['line_number']}")
    print(f"  Formatted: {phone['formatted']}")
    print()
```

### Assignment 4: Dynamic Data Pipeline with Lambdas & Custom Sorting

```python
def process_dataset(dataset):

    parsed_data = []
    
    for item in dataset:
        name = item[0]
        price_str = item[1]
        rating_str = item[2]
        
        price_parts = price_str.split(" ")
        price = float(price_parts[-1]) 
        
        rating_parts = rating_str.split(" ")
        rating = float(rating_parts[-1])  
        
        parsed_data.append((name, price, rating))
    
    filtered_data = filter(lambda x: x[1] <= 1000.0, parsed_data)

    mapped_data = map(lambda x: {
        "product": x[0],
        "price": x[1],
        "score": x[2]
    }, filtered_data)
 
    result = sorted(mapped_data, key=lambda x: x["score"], reverse=True)
   
    return result


data_input = [
    ("Laptop", "Price: 1200", "Rating: 4.8"),
    ("Phone", "Price: 800", "Rating: 4.5"),
    ("Mouse", "Price: 25", "Rating: 4.7"),
    ("Charger", "Price: 15", "Rating: 4.2")
]

result = process_dataset(data_input)

for item in result:
    print(item)
```

### Assignment 5: Stateful Ledger Scope Machine (LEGB Scopes & Closures)

```python

AUDIT_TRANSACTION_COUNT = 0

def create_bank_account(owner_name, initial_balance):
    """
    Creates a bank account with deposit, withdraw, and statement functions.
    
    Args:
        owner_name (str): Name of the account owner
        initial_balance (float): Starting balance
    
    Returns:
        dict: Dictionary containing the three functions
    """
    
   
    balance = initial_balance 
    history = []  
    
    history.append(f"Account created with {initial_balance}")
    
    def deposit(amount):
        """
        Adds money to the account.
        
        Args:
            amount (float): Amount to deposit
        """

        nonlocal balance, history
        
        balance += amount
       
        history.append(f"deposit {amount}")
        

        global AUDIT_TRANSACTION_COUNT
        AUDIT_TRANSACTION_COUNT += 1
    
    def withdraw(amount):
        """
        Takes money from the account if sufficient balance exists.
        
        Args:
            amount (float): Amount to withdraw
        
        Raises:
            ValueError: If balance is insufficient
        """
  
        nonlocal balance, history

        if balance >= amount:

            balance -= amount
    
            history.append(f"withdraw {amount}")
            
            global AUDIT_TRANSACTION_COUNT
            AUDIT_TRANSACTION_COUNT += 1
        else:
       
            raise ValueError("Insufficient balance")

    def get_statement():
        """
        Returns account information.
        
        Returns:
            tuple: (owner_name, current_balance, history_copy)
        """

        history_copy = history.copy()
        

        return (owner_name, balance, history_copy)

    return {
        "deposit": deposit,
        "withdraw": withdraw,
        "statement": get_statement
    }



print("Initial AUDIT_TRANSACTION_COUNT:", AUDIT_TRANSACTION_COUNT)
print()


acc = create_bank_account("Arham", 1000.0)
print("Account created for Arham with 1000.0")
print()

acc["deposit"](200.0)
print("Deposited 200.0")


acc["withdraw"](150.0)
print("Withdrew 150.0")
print()

print("Attempting to withdraw 2000.0...")
try:
    acc["withdraw"](2000.0)
except ValueError as e:
    print("Error:", e)  
print()

owner, bal, txn_history = acc["statement"]()
print("Owner:", owner)      
print("Balance:", bal)      
print("Transaction History:", txn_history)
print()

print("Final AUDIT_TRANSACTION_COUNT:", AUDIT_TRANSACTION_COUNT)
```

### Assignment 6: Server Log Analyzer & Traffic Classifier (Advanced RegEx)

```python
import re

def analyze_server_logs(logs_text):
    """
    Parses server logs and filters out local network requests.
    
    Args:
        logs_text (str): Multi-line string containing server logs
    
    Returns:
        list: List of dictionaries with external request details
    """
    
 
    pattern = re.compile(
        r'(?P<ip>\d+\.\d+\.\d+\.\d+)'  
        r' - - '                     
        r'\[(?P<time>[^\]]+)\]'         
        r' \"'                         
        r'(?P<method>[A-Z]+)'        
        r' '                           
        r'(?P<resource>/[^\s]*)'        
        r' '                            
        r'(?P<version>HTTP/\d+\.\d+)'   
        r'\"'                           
        r' '                            
        r'(?P<status>\d+)'              
        r' '                            
        r'(?P<bytes>\d+)'              
    )
    
   
    lines = logs_text.split("\n")
   
    results = []
   
    for line in lines:
       
        if line.strip() == "":
            continue
        
        match = pattern.match(line)
        
        if match is None:

            print(f"Warning: Could not parse line: '{line}'. Skipping.")
            continue 
        
        ip = match.group("ip")
        time = match.group("time")
        method = match.group("method")
        resource = match.group("resource")
        status = int(match.group("status"))  
        bytes_sent = int(match.group("bytes"))  
        
        
        is_local = False
        
        if ip.startswith("192.168."):
            is_local = True
        elif ip.startswith("10."):
            is_local = True
      
        if is_local:
            continue 
        
        log_entry = {
            "ip": ip,
            "time": time,
            "method": method,
            "resource": resource,
            "status": status,
            "bytes": bytes_sent
        }
     
        results.append(log_entry)

    return results


log_data = """192.168.1.5 - - [28/Aug/2026:10:00:00] "GET /index.html HTTP/1.1" 200 1024
8.8.8.8 - - [28/Aug/2026:10:10:00] "GET /api/v1/users HTTP/1.1" 200 4096
Corrupted log entry here
10.0.0.12 - - [28/Aug/2026:10:15:00] "POST /submit_data HTTP/1.1" 403 512
172.16.0.4 - - [28/Aug/2026:10:20:00] "POST /login HTTP/1.1" 401 256"""

result = analyze_server_logs(log_data)

print("\n=== External Requests ===")
for entry in result:
    print(entry)
```
