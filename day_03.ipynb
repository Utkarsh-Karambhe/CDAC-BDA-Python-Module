### Exercise 2: Movie Night Playlist

```python
playlist = ["Inception", "The Matrix", "Interstellar"]
usr_input = input("Enter the movie name")

if usr_input in playlist:
    print("Already Added!")
    playlist.sort()
    print(f"Alphabetic Playlist: {playlist}")
else:
    playlist.insert(0,usr_input)
    playlist.sort()
    print(f"Alphabetic Playlist: {playlist}")
```

### Exercise 4: Nightclub VIP Queue

```python
vip = ['Guido', 'Esha', 'Rajan', 'Kishori']

while True:
    guest_name = input("Enter the guest name: ")

    if(guest_name == "exit"):
        break

    if(guest_name in vip):
        i = vip.index(guest_name)
        element = vip.pop(i)
        vip.insert(0,element)

        print(f"{element} moved to the front.")
    else:
        print("Acess denied. Not in the VIP list.")

    print(f"Current VIP list: {vip}")
```

### Exercise 5: The Spy's Word Reverser

```python
sentence = input("Enter the message: ")

txt = sentence.split()

rev = [txt[i][::-1] for i in range(len(txt))]

result = " ".join(rev)
print(result)
```

### Exercise 6: Grading on a Curve


```python
n = int(input("Enter the number of marks you wish to enter: "))
scores = []
for i in range(n):
    score = int(input("Enter the number: "))
    scores.append(score)

original = scores[:]

for i in range(n):
    if(scores[i]<50):
        scores[i]+=10
    elif(scores[i]>=50):
        scores[i]+=5

for i in range(n):
    if(scores[i]<=100):
        continue
    else:
        scores.pop(i)
        scores.insert(i,100)


print(f"Original: {original}")
print(f"Curved: {scores}")
```

### Exercise 7: Treasure Map Coordinate Filter

```python
coords = [[12, 5], [-3, 14], [8, -2], [15, 9], [-5, -6]]

valid = []

for i in range(len(coords)):
    if(coords[i][0]>0 and coords[i][1]>1):
        valid.append(coords[i])

print(valid)
```

### Exercise 8: De-duplicating Shopping Cart

```python
cart = ["apple", "banana", "apple", "orange", "banana", "banana"]
cleaned = []

for item in cart:
    if item not in cleaned:
        cleaned.append(item)

print(f"Cleaned: {cleaned}")
```

### Exercise 9: The Josephus Elimination Game

```python

N = int(input("Enter number of soldiers (N): "))
K = int(input("Enter elimination interval (K): "))

soldiers = []
for i in range(1, N + 1):
    soldiers.append(i)

print(f"Soldier circle initialized: {soldiers}")

current_index = 0

while len(soldiers) > 1:
    current_index = (current_index + K - 1) % len(soldiers)

    eliminated = soldiers.pop(current_index)
    
    print(f"Eliminated soldier: {eliminated} (Remaining: {soldiers})")
print(f"The sole survivor is: {soldiers[0]}")
```
